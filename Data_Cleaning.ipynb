```python
# Remove duplicate order IDs
df = df.drop_duplicates(subset=["Order_ID"])
df
```

```python
#Check Missing Values
df.isnull().sum()
```

```python
# Find negative value
df = df.copy()
# Ensure Quantity, Price, and Total are numeric before comparison
df['Quantity'] = pd.to_numeric(df['Quantity'], errors='coerce')
df['Price'] = pd.to_numeric(df['Price'], errors='coerce')
df['Total'] = pd.to_numeric(df['Total'], errors='coerce')

# Find rows with negative values
negative_values = df[
    (df['Quantity'] < 0) |
    (df['Price'] < 0) |
    (df['Total'] < 0)
]

print(negative_values)
```

```python
# Handle mixed date formats
df = df.copy()
df["Order_Date"] = pd.to_datetime(df["Order_Date"], errors="coerce", format="mixed", dayfirst=True)
df = df[df["Order_Date"].notna()].copy()
df["Order_Date"] = df["Order_Date"].dt.strftime("%d-%m-%Y")

df
```

```python
# Clean category column

df[" Category"] = df[" Category"].astype(str).str.strip().str.lower()

category_map = {"sport": "Sports","electronic": "Electronics"}
df[" Category"] = df[" Category"].replace(category_map)
product_to_category = {"Laptop": "Electronics","Smartphone": "Electronics","Vacuum": "Home","Shoes": "Clothing","Jeans": "Clothing","Biography": "Books",
                       "Headphones":"Electronics","Basketball": "Sports","Yoga Mat":"Sports"}
df[" Category"] = df.apply(lambda row: product_to_category.get(row["Product"], row[" Category"]),axis=1)
df[" Category"] = df[" Category"].str.title()

df
```

```python
# Clean quantity column
# make sure Quantity is numeric
df["Quantity"] = pd.to_numeric(df["Quantity"], errors="coerce")

# replace negative values with NaN
df.loc[df["Quantity"] < 0, "Quantity"] = np.nan

# fill missing values with mode
mode_value = df["Quantity"].mode()[0]
df["Quantity"] = df["Quantity"].fillna(mode_value)
df
```

```python
# Clean price column
# Remove extreme price outlier
df = df[~(df['Price'] > 5000)].copy()

# Combine string cleaning and numeric conversion into one step
cleaned_price_series = df['Price'].astype(str).str.replace(r'[^0-9.]', '', regex=True)
df.loc[:, 'Price'] = pd.to_numeric(cleaned_price_series, errors='coerce')


# Fill missing values and round
df.loc[:, 'Price'] = df['Price'].fillna(df['Price'].mean()).round(2)
df
```

```python
# Create total column
df["Total"] = (df["Quantity"] * df["Price"]).round(2)
df
```
